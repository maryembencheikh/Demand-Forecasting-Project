# Imports

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

# Data exploration

```python
plt.figure(figsize=(10, 6))
sns.set(style="whitegrid")

df = pd.read_csv('demand_forecasting.csv') 
df.head() # Replace 'data.csv' with your actual data file
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df.dtypes
df.info()
df.isna().sum()
df.duplicated().sum()
```

```python
df.describe().T
```

```python
df.describe(include='object').T
```

```python
df.head(2)
```

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df['Weekday'] = df['Date'].dt.weekday
df
```

```python
df.columns
```

```python
df["Discounter Price"]=df["Price"]*(1-df["Discount"]/100)
df
```

```python
df["Sell Though Rate"] = df["Units Sold"] / df["Inventory Level"]
```

```python
df["Sell Though Rate"].describe()
```

```python
df.groupby("Category")["Demand"].agg(["mean","sum", "std"]).sort_values(by="mean", ascending=False).reset_index()
```

```python
df.groupby(["Region","Seasonality"])["Demand"].mean()
```

```python
df.groupby("Promotion")["Demand"].mean()
```

```python
df.pivot_table(values="Demand", index="Month", columns="Category", aggfunc="mean")
```

```python
sns.histplot(df["Demand"],bins=20, kde=True)
plt.title("Distribution of Demand")
```

```python
sns.scatterplot(data=df,x="Inventory Level",y="Units Sold")
plt.title("Inventory Level vs Units Sold")
plt.show()
```

```python
sns.boxplot(data=df,x="Category",y="Demand")
plt.xticks(rotation=45)
plt.title("Demand Distribution by Category")
plt.show()
```

```python
sns.boxplot(data=df,x="Weather Condition",y="Demand")
plt.xticks(rotation=45)
plt.title("Demand Distribution by Weather Condition")
plt.show()
```

```python
monthly_demand = df.groupby("Month")["Demand"].mean().reset_index()
```

```python
monthly_demand.plot(kind="bar")
plt.title("Average Monthly Demand")
plt.show()
```

```python
daily_demand=df.groupby("Date")["Demand"].sum()
daily_demand
```

```python
daily_demand.plot()
plt.title("Daily Demand Over Time")
plt.xlabel("Date")
plt.ylabel("Demand")
plt.show()
```

```python
sns.barplot(data=df,x="Promotion",y="Demand")
plt.title("Demand Distribution by Promotion")
plt.show()
```

```python
sns.regplot(data=df,x="Discounter Price",y="Demand")
plt.title("Discounter Price vs Demand")
plt.show()
```

```python
df.groupby("Seasonality")["Demand"].mean().plot(kind="bar",title="Demand by Seasonality")
plt.show
```

```python
df.groupby("Epidemic")["Demand"].mean().plot(kind="bar",title="Epidemic Impact on Demand")
plt.show()
```
